```python
# API GEOCODING DES VILLES / ADRESSE APRES FILTRE (récupéré coordonée)

def API_adresse(adresse_postale):
  url = 'https://data.geopf.fr/geocodage/search/?q='
  link = url + adresse_postale.replace(" ","+")
  r = requests.get(link).json()
  return r['features'][0]['geometry']['coordinates'][::-1]
```
