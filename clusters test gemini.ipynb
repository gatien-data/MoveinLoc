```python
import re

#Nettoyage de de Open en enlevant les tirets dans les villes, mise en int les CP pour retirer les float et création adresse complête

def normaliser(texte):
    """Remplace tirets/espaces multiples par un seul espace, en minuscule"""
    if pd.isna(texte):
        return texte
    texte = texte.lower()
    texte = re.sub(r'[-\s]+', ' ', texte)  # tirets ET espaces -> un espace
    return texte.strip()

def supprimer_ville(row):
    adresse = row['adresse']
    ville = row['ville']
    
    if pd.isna(adresse) or pd.isna(ville) or ville == '':
        return adresse
    
    ville_norm = normaliser(ville)
    
    # On construit un pattern qui accepte espace OU tiret entre les mots de la ville
    mots = ville_norm.split(' ')
    pattern = r'[\s-]+'.join(re.escape(m) for m in mots)
    
    adresse_nettoyee = re.sub(pattern, '', adresse, flags=re.IGNORECASE)
    
    # Nettoyage résiduel
    adresse_nettoyee = re.sub(r'\s{2,}', ' ', adresse_nettoyee)
    adresse_nettoyee = adresse_nettoyee.strip(' ,')
    
    return adresse_nettoyee

df_open['adresse'] = df_open.apply(supprimer_ville, axis=1)

df_open["cp"] = df_open["cp"].astype("Int64")
df_open["cp"] = df_open["cp"].astype(str)
df_open["adresse_complete"] = df_open["adresse"].fillna("") + " " + df_open["cp"].fillna("") + " " + df_open["ville"].fillna("")
```

```python
from collections import defaultdict

def detecter_theme(texte):

    texte = str(texte).lower()

    scores = defaultdict(int)

    for theme, mots_cles in themes_keywords.items():

        for mot in mots_cles:

            if mot in texte:
                scores[theme] += 1

    if not scores:
        return "autre"

    meilleur_theme = max(scores, key=scores.get)

    return meilleur_theme
df['texte_complet'] = (df['nom_event'].fillna('') + ' ' + df['desc'].fillna('') + ' ' + df['keyword'].fillna('')).str.lower()
df['theme'] = df['texte_complet'].apply(detecter_theme)
```

```python
import time

THEMES = list(themes_keywords.keys()) + ["autre"]

def detecter_theme_llm(nom_event, desc="", keyword=""):

    prompt = f"""
Tu regardes le nom_event de toutes les lignes et s'il s'agit d'un artiste (chanteur par exemple), tu remplaces le theme actuel par Concert.

{', '.join(THEMES)}

Nom :
{nom_event}

Description :
{desc}

Mots-clés :
{keyword}

Réponds uniquement par le nom exact du thème.
"""

    for tentative in range(5):

        try:

            response = client.models.generate_content(
                model="gemini-2.5-flash",
                contents=prompt
            )

            theme = response.text.strip()

            if theme in THEMES:
                return theme

            return "autre"

        except Exception as e:

            print(f"Tentative {tentative+1} échouée : {e}")

            # attente avant de réessayer
            time.sleep(5)

    return "autre"
```
