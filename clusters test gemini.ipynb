```python
import pandas as pd
```

```python
import re

#Nettoyage de df Open en enlevant les tirets dans les villes, mise en int les CP pour retirer les float et création adresse complête

def normaliser(texte):
    """Remplace tirets/espaces multiples par un seul espace, en minuscule"""
    if pd.isna(texte):
        return texte
    texte = texte.lower()
    texte = re.sub(r'[-\s]+', ' ', texte)  # tirets ET espaces -> un espace
    return texte.strip()

def supprimer_ville(row):
    adresse = row['adresse']
    ville = row['ville']
    
    if pd.isna(adresse) or pd.isna(ville) or ville == '':
        return adresse
    
    ville_norm = normaliser(ville)
    
    # On construit un pattern qui accepte espace OU tiret entre les mots de la ville
    mots = ville_norm.split(' ')
    pattern = r'[\s-]+'.join(re.escape(m) for m in mots)
    
    adresse_nettoyee = re.sub(pattern, '', adresse, flags=re.IGNORECASE)
    
    # Nettoyage résiduel
    adresse_nettoyee = re.sub(r'\s{2,}', ' ', adresse_nettoyee)
    adresse_nettoyee = adresse_nettoyee.strip(' ,')
    
    return adresse_nettoyee

df_open['adresse'] = df_open.apply(supprimer_ville, axis=1)

df_open["cp"] = df_open["cp"].astype("Int64")
df_open["cp"] = df_open["cp"].astype(str)
df_open["adresse_complete"] = df_open["adresse"].fillna("") + " " + df_open["cp"].fillna("") + " " + df_open["ville"].fillna("")
```
