```python
import re
import pandas as pd

def normaliser(texte):
    """Remplace tirets/espaces multiples par un seul espace, en minuscule"""
    if pd.isna(texte):
        return texte
    texte = texte.lower()
    texte = re.sub(r'[-\s]+', ' ', texte)  # tirets ET espaces -> un espace
    return texte.strip()

def supprimer_ville(row):
    adresse = row['adresse']
    ville = row['ville']
    
    if pd.isna(adresse) or pd.isna(ville) or ville == '':
        return adresse
    
    ville_norm = normaliser(ville)
    
    # On construit un pattern qui accepte espace OU tiret entre les mots de la ville
    mots = ville_norm.split(' ')
    pattern = r'[\s-]+'.join(re.escape(m) for m in mots)
    
    adresse_nettoyee = re.sub(pattern, '', adresse, flags=re.IGNORECASE)
    
    # Nettoyage résiduel
    adresse_nettoyee = re.sub(r'\s{2,}', ' ', adresse_nettoyee)
    adresse_nettoyee = adresse_nettoyee.strip(' ,')
    
    return adresse_nettoyee

df_open['adresse'] = df_open.apply(supprimer_ville, axis=1)

df_open["cp"] = df_open["cp"].astype("Int64")
df_open["cp"] = df_open["cp"].astype(str)
df_open["adresse_complete"] = df_open["adresse"].fillna("") + " " + df_open["cp"].fillna("") + " " + df_open["ville"].fillna("")
```

```python
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import KMeans

# --- Étape 1 : texte brut ---
df['texte_complet'] = df['nom_event'].fillna('') + ' ' + df['desc'].fillna('')

# --- Étape 2 : stop words (sans stemming) ---
stop_words_fr = stopwords.words('french') + ['the', 'cette', 'ans', '2026', 'saint', 'tout']

# --- Étape 3 : TF-IDF directement sur le texte original ---
vectorizer = TfidfVectorizer(
    max_features=1000,
    stop_words=stop_words_fr,
    ngram_range=(1, 2),
    min_df=2,
    max_df=0.3   # limite le poids des mots trop fréquents comme "marché"
)
X = vectorizer.fit_transform(df['texte_complet'])

# --- Étape 4 : KMeans ---
kmeans = KMeans(n_clusters=11, random_state=42, n_init=10)
df['cluster'] = kmeans.fit_predict(X)

# --- Étape 5 : interpréter les clusters ---
terms = vectorizer.get_feature_names_out()
order_centroids = kmeans.cluster_centers_.argsort()[:, ::-1]

for i in range(10):
    top_terms = [terms[ind] for ind in order_centroids[i, :10]]
    print(f"Cluster {i}: {', '.join(top_terms)}")
```

```python
themes_keywords = {
    'marché de Noël': ['marché de noël', 'marché noël'],
    'marché': ['marché','marché des producteurs', 'marché producteur','marché hebdomadaire','local','bio','locaux','producteurs','producteur', 'hebdomadaire','legumes','fermier','artisanat' ],
    'brocante / vide-grenier': [ 'bourse', 'anciens', 'vêtements', 'livres', 'collections','brocante', 'vide-grenier', 'vide grenier', 'vide-greniers', 'greniers', 'vide greniers', 'vide', 'greniers', 'marché puces', 'puces', 'brocante', 'brocante'],
    'fête nationale': ['feu d\'artifice', 'feux d\'artifice', '14 juillet', 'fête nationale', 'feu', 'artifice','nationale'],
    'fête patronale': ['fête patronale', 'fête votive', 'fête locale', 'ducasse', 'fête du village'],
    'concert': ['concert', 'live', 'chanteur', 'groupe', 'rock', 'pop', 'soirée dansante','gospel'],
    'festival': ['festival'],
    'visite guidée': ['visite guidée', 'balade commentée', 'découvrez', 'découverte'],
    'exposition': ['exposition', 'vernissage', 'galerie'],
    'théâtre': ['théâtre', 'spectacle', 'comédie', 'comedy'],
    'enfants':['petits', 'histoires', 'lectures', 'petits grands', 'enfants','famille'],
    'atelier': ['atelier', 'workshop', 'stage'],
    'foire': ['foire']
    }
```

```python
def detecter_theme(texte):
    texte = str(texte).lower()
    for theme, mots_cles in themes_keywords.items():
        if any(mot in texte for mot in mots_cles):
            return theme
    return 'autre'

df['texte_complet'] = (df['nom_event'].fillna('') + ' ' + df['desc'].fillna('')).str.lower()
df['theme'] = df['texte_complet'].apply(detecter_theme)
```
