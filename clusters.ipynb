```python
from collections import defaultdict

def detecter_theme(texte):

    texte = str(texte).lower()

    scores = defaultdict(int)

    for theme, mots_cles in themes_keywords.items():

        for mot in mots_cles:

            if mot in texte:
                scores[theme] += 1

    if not scores:
        return "autre"

    meilleur_theme = max(scores, key=scores.get)

    return meilleur_theme
df['texte_complet'] = (df['nom_event'].fillna('') + ' ' + df['desc'].fillna('') + ' ' + df['keyword'].fillna('')).str.lower()
df['theme'] = df['texte_complet'].apply(detecter_theme)
```

```python
from google import genai

client = genai.Client(api_key=GOOGLE_API_KEY)
```

```python
import re

def detecter_artistes(liste_evenements):

    texte = "\n".join(
        f"{i};{nom}"
        for i, nom in enumerate(liste_evenements)
    )

    prompt = f"""
Pour chaque ligne, réponds uniquement par :

numéro;OUI
ou
numéro;NON

Un artiste, un groupe de musique, un DJ, un chanteur, une chanteuse, un orchestre ou une chorale = OUI.

Tout le reste = NON.

Exemple :

0;OUI
1;NON
2;NON
3;OUI
"""

    response = client.models.generate_content(
        model="gemini-2.5-flash",
        contents=prompt
    )

    numeros = []

    for ligne in response.text.splitlines():

        ligne = ligne.strip()

        if re.fullmatch(r"\d+", ligne):
            numeros.append(int(ligne))

    return numeros
```

```python
import time

batch_size = 1000

indices = df.index.tolist()

for debut in range(0, len(indices), batch_size):

    batch = indices[debut:debut+batch_size]

    noms = df.loc[batch, "nom_event"].tolist()

    try:

        artistes = detecter_artistes(noms)

        for i in artistes:
            df.at[batch[i], "theme"] = "Concert"

        print(f"{debut+len(batch)} lignes traitées")

    except Exception as e:
        print(e)

    time.sleep(12)
```

```python
import time

THEMES = list(themes_keywords.keys()) + ["autre"]

def detecter_theme_llm(nom_event, desc="", keyword=""):

    prompt = f"""
Tu regardes le nom_event de toutes les lignes et s'il s'agit d'un artiste (chanteur par exemple), tu remplaces le theme actuel par Concert.

{', '.join(THEMES)}

Nom :
{nom_event}

Description :
{desc}

Mots-clés :
{keyword}

Réponds uniquement par le nom exact du thème.
"""

    for tentative in range(5):

        try:

            response = client.models.generate_content(
                model="gemini-2.5-flash",
                contents=prompt
            )

            theme = response.text.strip()

            if theme in THEMES:
                return theme

            return "autre"

        except Exception as e:

            print(f"Tentative {tentative+1} échouée : {e}")

            # attente avant de réessayer
            time.sleep(5)

    return "autre"
```
